# working with dates

```python
date['date'] = pd.to_datetime(date['date'])
```

```python
date.info()
```

1. extract year


```python
date['date_year'] = date['date'].dt.year
```

```python
date.sample(5)
```

2. Extract month


```python
date['date_month_no'] = date['date'].dt.month
date.sample(5)
```

```python
date['date_month_name'] = date['date'].dt.month_name()
date.sample(5)
```

3. Extract date

```python
date['date_now'] = date['date'].dt.day
date.head()
```

```python
# day of week
date['date_DoW'] = date['date'].dt.dayofweek
date.sample(5)
```

```python
# name of day
date['date_DoW_name'] = date['date'].dt.day_name()

date.sample(5)
```

```python
# is weekend?
date['date_is_weekend'] = date['date_DoW_name'].isin(['Sunday','Saterday']).astype('int')

date.sample(5)
```

Extract week of the year

```python
date['date_week'] = date['date'].dt.isocalendar().week
date.sample(5)
```

```python
date.sample(5)
```

 Extract Quarter

```python
date['Quarter'] = date['date'].dt.quarter
date.sample(5)
```

```python
# Extract Semester
date['Semester'] = np.where(date['Quarter'].isin([1,2]), 1, 2)
date.sample(5)
```

Extract Time elasped between dates

```python
import datetime 
today = datetime.datetime.now()

today
```

```python
today - date['date']
```

```python
(today - date['date']).dt.days
```

```python
# months passed
np.round((today - date['date']).dt.days / 30)
```
